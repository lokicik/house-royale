```python
import pandas as pd
import category_encoders as ce
import os

# 1. Veriyi Oku
df = pd.read_csv("../scraping/data/no_beylikdüzü.csv")

# 2. Bina Yaşı Temizliği
def yas_temizle(yas):
    if pd.isna(yas): return yas
    yas = str(yas).lower()
    if 'sıfır' in yas or 'yeni' in yas or yas == '0': return 0
    if 'yaşında' in yas: return float(yas.split()[0])
    if '5-10' in yas: return 7.5
    if '11-15' in yas: return 13
    if '16-20' in yas: return 18
    if '21 ve üzeri' in yas: return 25
    try:
        return float(yas)
    except:
        return None

df['bina_yasi'] = df['bina_yasi'].apply(yas_temizle)
df['bina_yasi'] = df['bina_yasi'].fillna(df['bina_yasi'].median())

# 3. Isıtma için One-Hot Encoding (dtype=int ile True/False yerine 1/0 yapar)
df = pd.get_dummies(df, columns=['isitma'], drop_first=True, dtype=int)

# 4. İlçe ve Mahalle için Target Encoding
# Tüm df'yi veriyoruz, sadece ilçe ve mahalle sütunlarını fiyata göre dönüştürüyor
target_encoder = ce.TargetEncoder(cols=['ilce', 'mahalle'])
df = target_encoder.fit_transform(df, df['fiyat'])

# 5. Dizin kontrolü ve Kaydetme
save_dir = '../scraping/data'
os.makedirs(save_dir, exist_ok=True)
save_path = f'{save_dir}/encoded.csv'

df.to_csv(save_path, index=False)
print(f"Veri başarıyla {save_path} konumuna kaydedildi!")
```

```python
import joblib

# Target encoder objesini geçici olarak scraping/data klasörüne kaydedelim
joblib.dump(target_encoder, '../scraping/data/target_encoder.pkl')
print("Target encoder başarıyla '../scraping/data/target_encoder.pkl' konumuna kaydedildi!")
```
